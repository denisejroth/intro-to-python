## 1. Checking your Python

If you've installed Python locally, you can check the version. (This cell works in Colab too — it
shows Colab's Python.)


````python
import sys
print("Python version:", sys.version)
````

You want **3.10 or higher**. If a local `python --version` in your terminal shows a `2.x`
version, use `python3` instead.


## 2. Where packages come from: `pip`

`pip` is Python's package installer. In Colab or a terminal you install a library like this:

```bash
pip install pandas
```

In a notebook cell, prefix it with `!` so the notebook runs it as a terminal command:


````python
# This works in Colab / Jupyter (the ! sends it to the terminal)
# !pip install pandas
print("Uncomment the line above to install a package.")
````

## 3. Virtual environments (local machine)

**The commands from the slides, collected here.** Run these in a *terminal*, not a notebook.

### Using `venv` (built into Python)

```bash
# 1. Create an environment named .venv inside your project folder
python -m venv .venv

# 2. Activate it
source .venv/bin/activate        # Mac / Linux
.venv\Scripts\activate           # Windows

# 3. Install packages — they now live inside .venv only
pip install pandas numpy

# 4. When done, leave the environment
deactivate
```

### Using `conda` (if you installed Anaconda / Miniconda)

```bash
conda create -n myenv python=3.11
conda activate myenv
conda install pandas numpy
conda deactivate
```


## 4. Recording your setup: `requirements.txt`

So others (or future-you) can recreate your environment exactly:

```bash
# Save the list of installed packages
pip freeze > requirements.txt

# Later, on another machine, recreate everything
pip install -r requirements.txt
```

This is the backbone of **reproducible** research code — anyone can rebuild your environment from
that one file.


## 5. A quick VS Code checklist

2. Install the **Python extension** (Extensions panel → search "Python")
3. **File → Open Folder** → pick your project folder
4. Create a `.py` file, or open a `.ipynb` notebook — VS Code handles both
5. Select your virtual environment as the interpreter (bottom-right, or Command Palette →
   "Python: Select Interpreter")


## 6. Cloning a repository from GitHub

To get a copy of a GitHub repository (like this course) onto your own machine, you **clone** it.
You only need the repo's URL — on its GitHub page, click the green **Code** button to find it.

### Terminal
```bash
# Navigate to where you want the repo, then:
git clone https://github.com/username/repo-name.git

# Later, to pull in the latest changes:
cd repo-name
git pull
```

### GitHub Desktop
*File → Clone Repository →* paste the URL → choose a local folder. Click **Fetch/Pull** later to
update.

### VS Code
Command Palette (`Ctrl/Cmd + Shift + P`) → type **"Git: Clone"** → paste the URL.

### Download ZIP (not really cloning)
Green **Code** button → **Download ZIP**. Quick, but you lose the history and can't easily pull
updates — fine for a one-off grab, not for ongoing work.
