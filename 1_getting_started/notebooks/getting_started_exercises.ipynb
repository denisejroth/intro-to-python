## 4. Arithmetic

Python is a calculator. Note the less obvious operators: `//` (floor division), `%`
(remainder), `**` (power).

```python
print(7 + 3)    # 10
print(7 / 3)    # 2.333...  (division always gives a float)
print(7 // 3)   # 2         (floor division: drops the decimal)
print(7 % 3)    # 1         (remainder / modulo)
print(7 ** 3)   # 343       (7 to the power of 3)
```

## 5. Strings and f-strings

Combine strings with `+`, or — much better — insert variables using **f-strings** (note the `f`
before the opening quote, and the `{}` around variables).

```python
first = "computational"
second = "text analysis"

# joining with +
print(first + " " + second)

# f-string: cleaner, and works with any type
field = "computational text analysis"
years = 3
print(f"I have studied {field} for {years} years.")
```

## 6. Converting between types

Change a value's type with `int()`, `float()`, `str()`. This matters because you **can't mix
types** carelessly.

```python
# "5" is text, not a number — converting lets us do math
text_number = "5"
print(int(text_number) + 3)   # 8

# going the other way
print(str(42) + " apples")    # "42 apples"
```
