## 1. Lists

An ordered, changeable collection in square brackets.

```python
words = ["cat", "dog", "bird"]
print(words)
print("Length:", len(words))
```

```python
print(words[0])    # first item — indexing starts at 0!
print(words[-1])   # last item
```

```python
words.append("fish")   # add to the end
words[0] = "lion"      # change an item
print(words)
```

> **✏️ Exercise 1**
>
> Make a list of three of your favorite foods. Print the first and last. Then append a fourth and print the whole list.

```python
# ✅ Solution
foods = ["pasta", "sushi", "curry"]
print(foods[0])
print(foods[-1])
foods.append("tacos")
print(foods)
```

## 2. Slicing

Grab a range with `[start:stop]` — the stop index is *not* included.

```python
nums = [10, 20, 30, 40, 50]
print(nums[1:3])   # [20, 30]
print(nums[:2])    # [10, 20]
print(nums[2:])    # [30, 40, 50]
```

> **✏️ Exercise 2**
>
> From `nums`, use slicing to get the middle three numbers (`[20, 30, 40]`).

```python
# ✅ Solution
print(nums[1:4])

# Comment: start at index 1 (the 20), stop *before* index 4 (the 50),
# so you get indices 1, 2, 3 = 20, 30, 40.
```

## 3. Dictionaries

Store **key → value** pairs; look things up by key.

```python
counts = {"cat": 3, "dog": 5}
print(counts["cat"])      # look up by key
counts["bird"] = 2        # add a new pair
counts["cat"] = 4         # update a value
print(counts)
```

```python
print(list(counts.keys()))     # all keys
print(list(counts.values()))   # all values
```

> **✏️ Exercise 3**
>
> Create a dictionary mapping three countries to their capital cities. Then look up and print one capital.

```python
# ✅ Solution
capitals = {"Netherlands": "Amsterdam", "France": "Paris", "Japan": "Tokyo"}
print(capitals["France"])

# Comment: dictionaries are perfect for lookups like this — the key ("France")
# takes you straight to its value, no searching required.
```

## 4. Tuples and sets

**Tuples** are unchangeable lists. **Sets** hold unique items.

```python
point = (52.1, 5.6)    # a tuple — round brackets, can't be changed
print(point[0])

tags = {"a", "b", "a", "c"}   # a set — duplicates removed
print(tags)
```

```python
# sets are great for finding unique items in a list
tokens = ["the", "cat", "sat", "the", "cat"]
unique = set(tokens)
print("Tokens:", len(tokens), "| Unique types:", len(unique))
```

> **✏️ Exercise 4**
>
> Given `letters = ['a', 'b', 'a', 'c', 'b', 'a']`, use a set to find how many *distinct* letters there are.

```python
# ✅ Solution
letters = ['a', 'b', 'a', 'c', 'b', 'a']
print(len(set(letters)))   # 3

# Comment: set() throws away duplicates, so set(letters) = {'a','b','c'},
# and len() of that is 3. This is exactly the tokens-vs-types idea.
```
