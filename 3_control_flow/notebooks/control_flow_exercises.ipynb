## 1. `if` / `elif` / `else`

Run code based on conditions. **Indentation matters** — the indented lines belong to the block.

```python
score = 72

if score >= 60:
    print("Pass")
else:
    print("Fail")
```

```python
if score >= 85:
    grade = "A"
elif score >= 70:
    grade = "B"
else:
    grade = "C"
print(grade)
```

## 2. `for` loops

Run code once for each item in a collection.

```python
words = ["cat", "dog", "bird"]
for word in words:
    print(word.upper())
```

```python
# range() loops a fixed number of times
for i in range(3):
    print(i)   # 0, 1, 2
```

```python
# enumerate() gives you position AND item
for i, word in enumerate(words):
    print(i, word)
```

## 3. Putting it together: counting words

Loops + conditions + a dictionary = a word counter.

```python
text = ["cat", "dog", "cat", "bird", "cat"]
counts = {}

for word in text:
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1

print(counts)
```
