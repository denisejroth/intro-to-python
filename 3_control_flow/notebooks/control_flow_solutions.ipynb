## 1. `if` / `elif` / `else`

Run code based on conditions. **Indentation matters** — the indented lines belong to the block.

```python
score = 72

if score >= 60:
    print("Pass")
else:
    print("Fail")
```

```python
if score >= 85:
    grade = "A"
elif score >= 70:
    grade = "B"
else:
    grade = "C"
print(grade)
```

> **✏️ Exercise 1**
>
> Write an `if`/`elif`/`else` that prints "child" if `age < 13`, "teenager" if `age < 20`, and "adult" otherwise. Test it with `age = 16`.

```python
# ✅ Solution
age = 16
if age < 13:
    print("child")
elif age < 20:
    print("teenager")
else:
    print("adult")

# Comment: order matters! Python checks top to bottom and stops at the first match.
# 16 is not < 13, but it is < 20, so "teenager" prints.
```

## 2. `for` loops

Run code once for each item in a collection.

```python
words = ["cat", "dog", "bird"]
for word in words:
    print(word.upper())
```

```python
# range() loops a fixed number of times
for i in range(3):
    print(i)   # 0, 1, 2
```

```python
# enumerate() gives you position AND item
for i, word in enumerate(words):
    print(i, word)
```

> **✏️ Exercise 2**
>
> Loop over `words` and print each word together with its length, like `cat: 3`.

```python
# ✅ Solution
for word in words:
    print(f"{word}: {len(word)}")

# Comment: combining a loop with an f-string is one of the most common patterns
# you'll write — process each item, format the result.
```

## 3. Putting it together: counting words

Loops + conditions + a dictionary = a word counter.

```python
text = ["cat", "dog", "cat", "bird", "cat"]
counts = {}

for word in text:
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1

print(counts)
```

> **✏️ Exercise 3**
>
> Using a loop, compute the **sum** of the numbers in `[4, 8, 15, 16, 23, 42]`. Start a variable at 0 and add each number to it. *(Then check with the built-in `sum()`.)*

```python
# ✅ Solution
numbers = [4, 8, 15, 16, 23, 42]
total = 0
for n in numbers:
    total += n
print(total)
print(sum(numbers))   # the built-in gives the same answer

# Comment: this 'accumulator' pattern — start at 0, add each item — is the manual
# version of what sum() does for you. Worth understanding before using the shortcut.
```
