## 2. Multiple arguments and defaults

```python
def power(base, exponent=2):
    return base ** exponent

print(power(5))          # 25 (exponent defaults to 2)
print(power(5, 3))       # 125
print(power(2, exponent=10))   # 1024
```

## 3. Scope

Variables created inside a function only exist there.

```python
def add(a, b):
    result = a + b   # local to this function
    return result

print(add(2, 3))   # 5
# print(result)    # this would ERROR — result doesn't exist out here
```

## 4. List comprehensions

A compact way to build a list from another.

```python
words = ["cat", "dog", "bird"]

# the loop way
lengths = []
for w in words:
    lengths.append(len(w))
print(lengths)

# the comprehension way — same result, one line
lengths = [len(w) for w in words]
print(lengths)
```

```python
# with a condition: keep only long words
long_words = [w for w in words if len(w) > 3]
print(long_words)
```
