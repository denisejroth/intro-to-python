## 1. Defining a function

`def`, a name, parameters, and `return`.

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Denise"))
print(greet("world"))
```

> **✏️ Exercise 1**
>
> Write a function `square(x)` that returns `x` multiplied by itself. Test it on 5 and 9.

```python
# ✅ Solution
def square(x):
    return x * x

print(square(5))   # 25
print(square(9))   # 81
```

## 2. Multiple arguments and defaults

```python
def power(base, exponent=2):
    return base ** exponent

print(power(5))          # 25 (exponent defaults to 2)
print(power(5, 3))       # 125
print(power(2, exponent=10))   # 1024
```

> **✏️ Exercise 2**
>
> Write a function `greet_formal(name, title="Dr.")` that returns `"Hello, Dr. Smith"`. Test it with the default title and with `title="Prof."`.

```python
# ✅ Solution
def greet_formal(name, title="Dr."):
    return f"Hello, {title} {name}"

print(greet_formal("Smith"))                # Hello, Dr. Smith
print(greet_formal("Jones", title="Prof."))  # Hello, Prof. Jones

# Comment: default arguments make functions flexible — callers can rely on the
# default or override it when they need to.
```

## 3. Scope

Variables created inside a function only exist there.

```python
def add(a, b):
    result = a + b   # local to this function
    return result

print(add(2, 3))   # 5
# print(result)    # this would ERROR — result doesn't exist out here
```

## 4. List comprehensions

A compact way to build a list from another.

```python
words = ["cat", "dog", "bird"]

# the loop way
lengths = []
for w in words:
    lengths.append(len(w))
print(lengths)

# the comprehension way — same result, one line
lengths = [len(w) for w in words]
print(lengths)
```

```python
# with a condition: keep only long words
long_words = [w for w in words if len(w) > 3]
print(long_words)
```

> **✏️ Exercise 3**
>
> Use a list comprehension to build a list of the **squares** of the numbers 1 to 5 (`[1, 4, 9, 16, 25]`). *(Hint: `range(1, 6)`.)*

```python
# ✅ Solution
squares = [n ** 2 for n in range(1, 6)]
print(squares)

# Comment: read it as 'n squared, for each n in 1..5'. Comprehensions are the
# idiomatic Python way to transform a sequence into a new list.
```

> **✏️ Exercise 4**
>
> Combine both skills: write a function `even_numbers(n)` that returns a list of all even numbers from 0 up to `n`, using a list comprehension inside. *(Hint: a number is even if `x % 2 == 0`.)*

```python
# ✅ Solution
def even_numbers(n):
    return [x for x in range(n + 1) if x % 2 == 0]

print(even_numbers(10))   # [0, 2, 4, 6, 8, 10]

# Comment: this packages a comprehension inside a reusable function — the two
# ideas from this module working together.
```
