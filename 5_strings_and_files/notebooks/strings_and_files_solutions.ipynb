## 1. String methods

Methods are functions attached to a value, called with a dot.

```python
text = "  Hello World  "
print(text.lower())
print(text.upper())
print(text.strip())            # trim surrounding spaces
print(text.replace("World", "Python"))
```

> **✏️ Exercise 1**
>
> Given `messy = "  POLICY Document  "`, clean it up: strip the spaces and make it lowercase, giving `"policy document"`.

```python
# ✅ Solution
messy = "  POLICY Document  "
clean = messy.strip().lower()
print(clean)   # "policy document"

# Comment: you can chain methods — .strip() returns a string, which .lower() then
# acts on. This left-to-right chaining is very common in text cleaning.
```

## 2. Splitting and joining

The heart of tokenization.

```python
sentence = "the cat sat"
words = sentence.split()      # ["the", "cat", "sat"]
print(words)

print(" ".join(words))         # "the cat sat"
print("-".join(words))        # "the-cat-sat"
```

> **✏️ Exercise 2**
>
> Take the sentence `"computational text analysis is fun"`, split it into words, and print how many words it has. Then join it back using underscores.

```python
# ✅ Solution
s = "computational text analysis is fun"
words = s.split()
print("Word count:", len(words))
print("_".join(words))

# Comment: split() then len() is the simplest possible word count — exactly what
# the text-analysis course starts with.
```

## 3. Checking and searching

```python
word = "policy"
print("pol" in word)           # True
print(word.startswith("pol"))  # True
print(word.endswith("cy"))     # True
print(word.count("o"))         # 1
```

> **✏️ Exercise 3**
>
> Given a list of words, use a loop (or comprehension) to keep only those that contain the letter `"a"`. Test on `["cat", "dog", "bird", "snake"]`.

```python
# ✅ Solution
animals = ["cat", "dog", "bird", "snake"]
with_a = [w for w in animals if "a" in w]
print(with_a)   # ["cat", "snake"]

# Comment: the 'in' keyword checks for a substring, and the comprehension keeps
# only the words where that check is True.
```

## 4. Reading and writing files

Use a `with` block — it closes the file automatically.

```python
# write a file
with open("notes.txt", "w") as f:
    f.write("Hello file\nSecond line")

# read it back
with open("notes.txt") as f:
    content = f.read()
print(content)
```

> **✏️ Exercise 4**
>
> Write a file called `shopping.txt` containing three items (one per line), then read it back and print it. *(Hint: `"\n"` is a line break.)*

```python
# ✅ Solution
with open("shopping.txt", "w") as f:
    f.write("apples\nbread\ncoffee")

with open("shopping.txt") as f:
    print(f.read())

# Comment: "\n" separates lines. The with-block guarantees the file is properly
# closed even if something goes wrong — always prefer it over open()/close().
```
