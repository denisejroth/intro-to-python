## 1. Creating arrays

An array is like a list, but built for fast math.

```python
arr = np.array([1, 2, 3, 4])
print(arr)
print("Shape:", arr.shape)   # (4,) = 4 elements
```

> Create a NumPy array holding the numbers 10, 20, 30, 40, 50. Print it and its shape.

```python
# ✅ Solution
arr = np.array([10, 20, 30, 40, 50])
print(arr)
print(arr.shape)   # (5,)
```

## 2. Vectorized math

Operations apply to the **whole array at once** — no loop.

```python
arr = np.array([1, 2, 3, 4])
print(arr * 2)    # [2 4 6 8]
print(arr + 10)   # [11 12 13 14]
print(arr ** 2)   # [1 4 9 16]
```

```python
# contrast with a plain list, where * 2 just repeats it
plain = [1, 2, 3, 4]
print(plain * 2)   # [1, 2, 3, 4, 1, 2, 3, 4] — NOT element-wise!
```

> **✏️ Exercise 2**
>
> You have temperatures in Celsius: `np.array([0, 20, 37, 100])`. Convert them all to Fahrenheit in one line. *(Formula: F = C × 9/5 + 32.)*

```python
# ✅ Solution
celsius = np.array([0, 20, 37, 100])
fahrenheit = celsius * 9/5 + 32
print(fahrenheit)   # [ 32.  68.  98.6 212. ]

# Comment: this is the beauty of vectorization — one expression transforms every
# element, no loop needed, and it's fast even for millions of numbers.
```

## 3. Summaries and filtering

```python
data = np.array([4, 8, 15, 16, 23])
print("Sum:", data.sum())
print("Mean:", data.mean())
print("Max:", data.max())
print("Std:", data.std())
```

```python
# boolean filtering: keep elements meeting a condition
print(data[data > 10])   # [15 16 23]
```

> **✏️ Exercise 3**
>
> Given `scores = np.array([45, 88, 72, 30, 95, 60])`, use filtering to (a) print only the passing scores (>= 60), and (b) print the mean of *all* scores.

```python
# ✅ Solution
scores = np.array([45, 88, 72, 30, 95, 60])
print(scores[scores >= 60])   # [88 72 95 60]
print(scores.mean())          # 65.0

# Comment: scores >= 60 creates a True/False mask, and indexing with it keeps only
# the True positions. This same idea powers pandas filtering in the next module.
```
