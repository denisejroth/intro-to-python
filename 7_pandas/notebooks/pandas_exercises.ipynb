## 1. Creating a DataFrame

A DataFrame is a table with named columns and indexed rows.

```python
df = pd.DataFrame({
    "city": ["Wageningen", "Utrecht", "Amsterdam", "Rotterdam"],
    "population": [40000, 360000, 900000, 650000],
    "region": ["Gelderland", "Utrecht", "N-Holland", "S-Holland"],
})
df
```

## 3. Filtering rows

Keep rows meeting a condition — the most common data task.

```python
# cities with more than 500,000 people
big = df[df["population"] > 500000]
big
```

```python
# combine conditions with & (and) / | (or) — note the parentheses!
df[(df["population"] > 100000) & (df["region"] != "Utrecht")]
```

## 4. New columns and grouping

```python
# create a new column
df["pop_thousands"] = df["population"] / 1000
df
```

## 5. A quick plot

pandas plots straight from the DataFrame.

```python
import matplotlib.pyplot as plt
df.plot(x="city", y="population", kind="bar", legend=False)
plt.title("Population by city")
plt.tight_layout()
plt.show()
```
